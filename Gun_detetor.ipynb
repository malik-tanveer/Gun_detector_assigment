```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
model.train(data="/content/Dataset/data.yaml", save=True, epochs=10, batch=64)
```

```python
predict = model.predict("images.jpg", save=True, conf=0.5)
```
